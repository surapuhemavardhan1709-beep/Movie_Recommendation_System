```python
import pandas as pd

movies = pd.read_csv("data/movies.csv")
credits = pd.read_csv("data/credits.csv")

print("Movies shape:", movies.shape)
print("Credits shape:", credits.shape)
```

```python
print("Movies columns:")
print(movies.columns)

print("\nCredits columns:")
print(credits.columns)
```

```python
movies = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id"
)

print("Merged dataset shape:", movies.shape)
```

```python
movies.head()
```

```python
movies = movies[['movie_id', 'title_x', 'overview', 'genres', 'keywords', 'cast', 'crew']]

movies.head()
```

```python
movies.rename(columns={'title_x': 'title'}, inplace=True)

movies.head()
```

```python
print(movies.isnull().sum())
```

```python
movies['overview'] = movies['overview'].fillna('')

print(movies.isnull().sum())
```

```python
import ast

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)

movies[['title', 'genres', 'keywords']].head()
```

```python
def convert3(text):
    L = []
    for i, item in enumerate(ast.literal_eval(text)):
        if i < 3:
            L.append(item['name'])
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)

movies[['title', 'cast']].head()
```

```python
def fetch_director(text):
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            return [i['name']]
    return []
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)

movies[['title', 'crew']].head()
```

```python
print("Title:", movies.iloc[0]['title'])
print("Overview:", movies.iloc[0]['overview'])
print("Genres:", movies.iloc[0]['genres'])
print("Keywords:", movies.iloc[0]['keywords'])
print("Cast:", movies.iloc[0]['cast'])
print("Director:", movies.iloc[0]['crew'])
```

```python
movies['tags'] = (
    movies['overview'] +
    movies['genres'].apply(lambda x: ' '.join(x)) +
    movies['keywords'].apply(lambda x: ' '.join(x)) +
    movies['cast'].apply(lambda x: ' '.join(x)) +
    movies['crew'].apply(lambda x: ' '.join(x))
)

movies[['title', 'tags']].head()
```

```python
movies['tags'] = movies['tags'].apply(lambda x: x.lower().replace(" ", ""))

movies[['title', 'tags']].head()
```

```python
movies['tags'] = movies['tags'].apply(
    lambda x: x.lower().replace(" ", "")
)

movies[['title', 'tags']].head()
```

```python
movies['tags'] = (
    movies['overview'] + ' ' +
    movies['genres'].apply(lambda x: ' '.join(x).replace(" ", "")) + ' ' +
    movies['keywords'].apply(lambda x: ' '.join(x).replace(" ", "")) + ' ' +
    movies['cast'].apply(lambda x: ' '.join(x).replace(" ", "")) + ' ' +
    movies['crew'].apply(lambda x: ' '.join(x).replace(" ", ""))
)

movies['tags'] = movies['tags'].apply(lambda x: x.lower())

movies[['title', 'tags']].head()
```

```python
movies = movies[['movie_id', 'title', 'tags']]

movies.head()
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
movies['tags'] = movies['tags'].apply(stem)

movies.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')

vectors = cv.fit_transform(movies['tags']).toarray()

print("Vectors shape:", vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity shape:", similarity.shape)
```

```python
def recommend(movie):
    index = movies[movies['title'] == movie].index[0]
    
    distances = similarity[index]
    
    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    for i in movies_list:
        print(movies.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
import pickle

pickle.dump(movies, open('movie_dict.pkl', 'wb'))
pickle.dump(similarity, open('similarity.pkl', 'wb'))

print("Files saved successfully!")
```
